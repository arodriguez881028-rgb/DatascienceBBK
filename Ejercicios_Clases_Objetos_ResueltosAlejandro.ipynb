## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
# 1. Implementar la clase "Tienda"
class Tienda:

    tipo = "Electrodomésticos"
    abierta = True
    
    def __init__(self, nombre: str, direccion: str, num_empleados: int, ventas_ultimos_meses: list):
       
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_ultimos_meses = ventas_ultimos_meses
        
    def calcular_ventas_totales(self) -> float:
        
        return sum(self.ventas_ultimos_meses)
        
    def media_ventas_por_empleado(self) -> float:
        
        if self.num_empleados == 0:
            return 0.0
        return self.calcular_ventas_totales() / self.num_empleados
        
    def datos_tienda(self) -> str:
        
        return f"{self.nombre}, {self.direccion}"
        
    def ventas_ultimo_mes(self) -> float:
        
        return self.ventas_ultimos_meses[-1]
        
    def proyeccion_marketing(self, inversion_marketing: float) -> list:
       
        factor = 1.2 if inversion_marketing < 1000 else 1.5
        self.ventas_ultimos_meses = [v * factor for v in self.ventas_ultimos_meses]
        return self.ventas_ultimos_meses


# 2. Crear tres tiendas con datos inventados
tienda_1 = Tienda("ElectroBilbao", "Gran Vía 45", 6, [12000, 15000, 18000])
tienda_2 = Tienda("TechNorte", "Avenida de la Libertad 12", 4, [8000, 9500, 11000])
tienda_3 = Tienda("MegaElectro", "Avenida Sabino Arana 8", 10, [22000, 25000, 29000])

lista_tiendas = [tienda_1, tienda_2, tienda_3]


# 3. Comprobar en al menos una de ellas todo lo implementado
print("=== COMPROBACIÓN EN TIENDA 1 ===")
print("Atributo de clase (tipo):", tienda_1.tipo)
print("Atributo de clase (abierta):", tienda_1.abierta)
print("Nombre y dirección:", tienda_1.datos_tienda())
print("Número de empleados:", tienda_1.num_empleados)
print("Ventas últimos 3 meses (originales):", tienda_1.ventas_ultimos_meses)
print("Total de ventas acumuladas:", tienda_1.calcular_ventas_totales(), "€")
print("Media de ventas por empleado:", round(tienda_1.media_ventas_por_empleado(), 2), "€")
print("Ventas del último mes:", tienda_1.ventas_ultimo_mes(), "€")

# Proyección de marketing invirtiendo 500€ (< 1000€ -> factor 1.2)
ventas_proyectadas = tienda_1.proyeccion_marketing(500)
print("Ventas tras proyección de marketing (500€ invertidos):", ventas_proyectadas)
print("Comprobación del atributo reescrito:", tienda_1.ventas_ultimos_meses)


# 4. Calcular las ventas del último mes de todas las tiendas usando un bucle for
print("\n=== VENTAS DEL ÚLTIMO MES (TODAS LAS TIENDAS) ===")
for tienda in lista_tiendas:
    print(f"Tienda: {tienda.nombre} -> Ventas último mes: {tienda.ventas_ultimo_mes()} €")


# 5. Imprimir los nombres de las tiendas cuya dirección lleve Avenida
print("\n=== TIENDAS EN AVENIDA ===")
for tienda in lista_tiendas:
    if "Avenida" in tienda.direccion:
        print(f"La tienda {tienda.nombre} está en una avenida: {tienda.direccion}")


# 6. Documentación de la clase
print("\n=== DOCUMENTACIÓN DE LA CLASE ===")
print(Tienda.__doc__)

```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
# 1. Implementar la clase Perro
class Perro:
   
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0
    
    def __init__(self, raza: str, pelo: str = "Marrón", dueno: str = None):
        """
        Constructor de la clase Perro.
        
        Args:
            raza (str): Raza del perro (obligatoria).
            pelo (str, opcional): Color del pelo. Por defecto "Marrón".
            dueno (str, opcional): Nombre del dueño. Por defecto None (sin dueño).
        """
        self.raza = raza
        self.pelo = pelo
        self.dueno = dueno
        
    def andar(self, aumento_velocidad: float):
        """Aumenta la velocidad del perro sumando el valor recibido a la velocidad previa."""
        self.velocidad += aumento_velocidad
        print(f"El perro acelera +{aumento_velocidad} km/h. Velocidad actual: {self.velocidad} km/h")
        
    def parar(self):
        """Detiene al perro restableciendo su velocidad a 0."""
        self.velocidad = 0
        print("El perro se ha parado. Velocidad actual: 0 km/h")
        
    def ladrar(self, mensaje: str) -> str:
        """Devuelve el ladrido concatenado con el mensaje de entrada."""
        return "GUAU!" + mensaje


# 2. Crear un objeto de tipo Perro, sin dueño
perro_sin_dueno = Perro(raza="Pastor Alemán")


# 3. Comprobar que están bien todos sus atributos
print("=== COMPROBACIÓN DE ATRIBUTOS DEL PERRO ===")
print("Raza:", perro_sin_dueno.raza)
print("Color de pelo (por defecto):", perro_sin_dueno.pelo)
print("Dueño (por defecto):", perro_sin_dueno.dueno)
print("Patas (atributo de clase):", perro_sin_dueno.patas)
print("Orejas (atributo de clase):", perro_sin_dueno.orejas)
print("Ojos (atributo de clase):", perro_sin_dueno.ojos)
print("Velocidad inicial (por defecto parado):", perro_sin_dueno.velocidad, "km/h")


# 4. Probar que ande y comprobar su velocidad
print("\n=== PROBANDO EL MÉTODO ANDAR ===")
perro_sin_dueno.andar(12)
perro_sin_dueno.andar(8)
print("Velocidad comprobada tras andar:", perro_sin_dueno.velocidad, "km/h")


# 5. Páralo
print("\n=== PROBANDO EL MÉTODO PARAR ===")
perro_sin_dueno.parar()
print("Velocidad comprobada tras parar:", perro_sin_dueno.velocidad, "km/h")


# Prueba extra: método ladrar
print("\n=== PROBANDO EL MÉTODO LADRAR ===")
print(perro_sin_dueno.ladrar(" Dame comida"))


# 6. Documentación de la clase Perro
print("\n=== DOCUMENTACIÓN DE LA CLASE ===")
print(Perro.__doc__)

```
