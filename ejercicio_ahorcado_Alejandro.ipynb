## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
def jugar_ahorcado(palabra_secreta):
    palabra = palabra_secreta.upper()
    letras_adivinadas = set()
    errores = 0
    max_errores = 6

    print("--- ¡BIENVENIDO AL JUEGO DEL AHORCADO! ---")

    while errores < max_errores:
        # Construimos el estado actual de la palabra con letras o guiones bajos
        estado_actual = [
            letra if letra in letras_adivinadas else "_" for letra in palabra
        ]
        print("\nPalabra:", " ".join(estado_actual))

        # Comprobamos si ya adivinó todas las letras
        if "_" not in estado_actual:
            print("\n¡Felicidades! Has adivinado la palabra completa.")
            return

        # Pedimos una letra al usuario
        letra_usuario = input("Introduce una letra: ").strip().upper()

        # Validaciones de entrada
        if len(letra_usuario) != 1 or not letra_usuario.isalpha():
            print("Por favor, introduce solo una letra válida.")
            continue

        if letra_usuario in letras_adivinadas:
            print(f"Ya habías probado la letra '{letra_usuario}'. Intenta con otra.")
            continue

        # Registramos la letra probada
        letras_adivinadas.add(letra_usuario)

        # Verificamos si la letra está en la palabra
        if letra_usuario in palabra:
            print(f"¡Bien hecho! La letra '{letra_usuario}' está en la palabra.")
        else:
            errores += 1
            print(f"La letra '{letra_usuario}' no está. Llevas {errores} error(es) de {max_errores}.")

    print("\nFin del juego")
    print(f"La palabra correcta era: {palabra}")


# Ejemplo de uso:
jugar_ahorcado("SISTEMAS")
````
