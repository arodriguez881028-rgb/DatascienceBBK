# Nombres ficticios

```python
import numpy as np
import pandas as pd
```

### Paso 2. Crea los tres DataFrames a partir de los siguientes datos originales.

```python
raw_data_1 = {
        'subject_id': ['1', '2', '3', '4', '5'],
        'first_name': ['Alex', 'Amy', 'Allen', 'Alice', 'Ayoung'], 
        'last_name': ['Anderson', 'Ackerman', 'Ali', 'Aoni', 'Atiches']}

raw_data_2 = {
        'subject_id': ['4', '5', '6', '7', '8'],
        'first_name': ['Billy', 'Brian', 'Bran', 'Bryce', 'Betty'], 
        'last_name': ['Bonder', 'Black', 'Balwner', 'Brice', 'Btisan']}

raw_data_3 = {
        'subject_id': ['1', '2', '3', '4', '5', '7', '8', '9', '10', '11'],
        'test_id': [51, 15, 15, 61, 16, 14, 15, 1, 61, 16]}
```

### Paso 3. Asigna cada DataFrame a una variable llamada data1, data2 y data3.

```python
data1= pd.DataFrame(raw_data_1)
data2= pd.DataFrame(raw_data_2)
data3= pd.DataFrame(raw_data_3)
```

### Paso 4. Une los dos primeros DataFrames por filas y asigna el resultado a all_data.

```python
all_data = pd.concat([data1, data2], ignore_index=True)
all_data
```

### Paso 5. Une los dos primeros DataFrames por columnas y asigna el resultado a all_data_col.

```python
all_data_col = pd.concat([data1, data2], axis=1)
all_data_col
```

### Paso 6. Muestra los valores únicos de data3.

```python
# Valores únicos de subject_id y test_id en data3
print('Valores únicos de subject_id:', data3['subject_id'].unique())
print('Valores únicos de test_id:', data3['test_id'].unique())
# O el número de valores únicos por columna:
data3.nunique()
```

### Paso 7. Combina all_data y data3 usando la columna subject_id.

```python
# Merge por defecto realiza inner join sobre la clave subject_id
pd.merge(all_data, data3, on='subject_id')
```

### Paso 8. Combina data1 y data2 conservando únicamente los registros cuyo subject_id aparece en ambos DataFrames.

```python
# Inner join entre data1 y data2 por subject_id
pd.merge(data1, data2, on='subject_id', how='inner')
```

### Paso 9. Combina todos los registros de data1 y data2, incluidos los que no coinciden en uno de los lados.

```python
# Outer join (full outer join) entre data1 y data2
pd.merge(data1, data2, on='subject_id', how='outer')
```
