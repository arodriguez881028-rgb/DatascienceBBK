### 12. ¿Cuántas profesiones diferentes hay? ¿Y personas por profesión?

```python
# Número de profesiones distintas y número de personas en cada profesión.
print('Profesiones diferentes:', users['occupation'].nunique())
users['occupation'].value_counts()
```

### 13. ¿Cuál es la profesión más común?

```python
# Profesión más común (incluye posibles empates).
frecuencias_profesion = users['occupation'].value_counts()
frecuencias_profesion[frecuencias_profesion == frecuencias_profesion.max()]
```

### 18. ¿Cuál es la edad menos frecuente?

```python
# Todas las edades con la frecuencia mínima observada.
frecuencias_edad = users['age'].value_counts()
frecuencias_edad[frecuencias_edad == frecuencias_edad.min()].sort_index()
```
