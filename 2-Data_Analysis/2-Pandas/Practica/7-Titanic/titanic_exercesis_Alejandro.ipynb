9. ¿Cómo se llamaba la persona de mayor edad que murió en el Titanic? ¿Viajaba sola o tenía familiares a bordo?

Para responder a la segunda pregunta, averigua qué columnas contienen esa información. Como analista de datos, parte del trabajo consiste en conocer el conjunto de datos. En este caso, las columnas relevantes son:

- `sibsp`: número de hermanos o cónyuges a bordo.
- `parch`: número de padres o hijos a bordo.

```python
# Filtramos por personas que murieron (survived == 0)
dead_people = titanic[titanic['survived'] == 0]
oldest_dead = dead_people.loc[dead_people['age'].idxmax()]
print('Nombre:', oldest_dead['name'])
print('Edad:', oldest_dead['age'])
print('sibsp (hermanos/cónyuges):', oldest_dead['sibsp'])
print('parch (padres/hijos):', oldest_dead['parch'])
if oldest_dead['sibsp'] == 0 and oldest_dead['parch'] == 0:
    print('Respuesta: Viajaba sola (sibsp=0 y parch=0).')
else:
    print('Respuesta: Tenía familiares a bordo.')
```

34. Usa [`kdeplot`](https://seaborn.pydata.org/generated/seaborn.kdeplot.html) de `seaborn` para representar la distribución por edad de las personas que no sobrevivieron.

Pista: primero filtra el conjunto `titanic` para quedarte con las filas en las que `survived` sea 0 y selecciona únicamente la columna `age`.

Pasa estos argumentos a la función:

- `color = "red"`
- `label = "Not Survived"`
- `shade = True`

El resultado debería parecerse a este:

![Gráfico kdeplot](./img/kdeplotsur.png)

```python
not_survived_age = titanic[titanic['survived'] == 0]['age']
plt.figure(figsize=(8, 5))
# Nota: shade=True está soportado o fill=True en versiones nuevas
try:
    sns.kdeplot(not_survived_age, color='red', label='Not Survived', fill=True)
except TypeError:
    sns.kdeplot(not_survived_age, color='red', label='Not Survived', shade=True)
plt.title('Distribución de edad de los no supervivientes')
plt.xlabel('Edad')
plt.legend()
plt.show()
```
