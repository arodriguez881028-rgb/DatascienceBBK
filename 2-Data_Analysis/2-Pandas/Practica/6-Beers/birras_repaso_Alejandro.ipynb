## Ejercicio 10
El amargor de las cervezas se mide mediante el nivel de IBU.
1. ¿Cuál es el nivel medio de amargor?
2. ¿Cuál es el máximo? ¿Y el mínimo?
3. ¿Qué cerveza tiene el nivel de amargor máximo? ¿Y cuál tiene el mínimo?

```python
# 1. Nivel medio de IBU
ibu_mean = df_beers['ibu'].mean()
print(f'IBU medio: {ibu_mean:.2f}')

# 2. Máximo y mínimo de IBU
ibu_max = df_beers['ibu'].max()
ibu_min = df_beers['ibu'].min()
print(f'IBU máximo: {ibu_max}, IBU mínimo: {ibu_min}')

# 3. Cerveza con amargor máximo y mínimo
cerveza_max = df_beers.loc[df_beers['ibu'].idxmax()]['name']
cerveza_min = df_beers.loc[df_beers['ibu'].idxmin()]['name']
print(f'Cerveza con IBU máximo: {cerveza_max}')
print(f'Cerveza con IBU mínimo: {cerveza_min}')
```

## Ejercicio 11
No está claro cuánto varían los valores de IBU. Utiliza una o dos métricas de estadística descriptiva, sin gráficos, para explicar la dispersión de esta variable.

```python
# Desviación típica (std) y Rango Intercuartílico (IQR)
std_ibu = df_beers['ibu'].std()
iqr_ibu = df_beers['ibu'].quantile(0.75) - df_beers['ibu'].quantile(0.25)
print(f'Desviación estándar de IBU: {std_ibu:.2f}')
print(f'Rango Intercuartílico (IQR) de IBU: {iqr_ibu:.2f}')
```

## Ejercicio 12
¿Existe una relación lineal entre las variables numéricas del conjunto de datos? Intenta mostrarlo mediante un gráfico.

```python
# Matriz de correlación y mapa de calor
corr = df_beers[['abv', 'ibu', 'ounces']].corr()
plt.figure(figsize=(6, 4))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Matriz de correlación')
plt.show()

# Dispersión entre abv e ibu
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_beers, x='abv', y='ibu')
plt.title('Relación entre Alcohol (abv) y Amargor (ibu)')
plt.show()
```

## Ejercicio 13
Representa en un histograma el grado de alcohol (columna abv). ¿La distribución es simétrica o asimétrica?

```python
plt.figure(figsize=(7, 4))
sns.histplot(df_beers['abv'].dropna(), kde=True, bins=30)
plt.title('Distribución del Grado de Alcohol (ABV)')
plt.xlabel('ABV')
plt.show()
print('Skewness (asimetría):', df_beers['abv'].skew())
print('Tiene asimetría positiva (cola a la derecha), con la mayoría concentrada entre 0.04 y 0.07.')
```

## Ejercicio 18
Combina los dos conjuntos de datos.

```python
# Cruzamos por 'brewery_id' mediante un inner join (o left join)
df_completo = pd.merge(df_beers, df_breweries, on='brewery_id')
df_completo.head()
```

## Ejercicio 19
1. ¿Cuántos estados distintos hay?
2. ¿Y cuántas ciudades distintas?
3. ¿Cuántas cervecerías hay?

## Ejercicio 20
Agrupa por estado y ciudad y calcula la media, la mediana y el valor máximo del grado de alcohol.

```python
resumen_abv = df_completo.groupby(['state', 'city'])['abv'].agg(['mean', 'median', 'max'])
resumen_abv
```

## Ejercicio 21

## Ejercicio 23
Crea un DataFrame con todas las cervezas de las cervecerías que tengan al menos una cerveza con un abv igual o superior a 0,07.

```python
# Identificamos las IDs de cervecerías que tienen al menos una cerveza con abv >= 0.07
cervecerias_fuertes = df_completo[df_completo['abv'] >= 0.07]['brewery_id'].unique()

# Filtramos el dataframe completo con esas cervecerías
df_cervezas_fuertes = df_completo[df_completo['brewery_id'].isin(cervecerias_fuertes)]
df_cervezas_fuertes
```

## Ejercicio 24
Calcula la media del amargor de todas las cervezas y obtén una tabla con las cervezas cuyos estados tengan una media de amargor superior a esa media.

```python
# 1. Media de amargor global
media_global_ibu = df_completo['ibu'].mean()
print(f'Media global de IBU: {media_global_ibu:.2f}')

# 2. Media de IBU por estado
media_por_estado = df_completo.groupby('state')['ibu'].transform('mean')

# 3. Filtrar cervezas cuyos estados superan la media global
cervezas_estados_amargos = df_completo[media_por_estado > media_global_ibu]
cervezas_estados_amargos
```
