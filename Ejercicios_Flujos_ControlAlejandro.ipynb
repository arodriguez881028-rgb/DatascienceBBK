## Ejercicio 3
En este ejercicio vamos a realizar un programa muy parecido. Para este caso queremos que se cumplan las siguientes condiciones:

1. Primero se compruebe si el distrito es "Retiro". Si es asá, que imprima "Distrito Retiro", y si no, "Otro distrito"
2. Si el distrito es "Retiro", comprueba si la superficie es mayor de 100 metros cuadrados. En tal caso, que imprima un precio de 1000, y si no, de 500.

Prueba a usar `ifs` anidados

```python
# Solicitamos los datos al usuario mediante la consola
distrito = input("Introduce el distrito de la casa: ")
superficie = float(input("Introduce la superficie en metros cuadrados: "))

# 1. Primero comprobamos si el distrito es "Retiro"
if distrito == "Retiro":
    print("Distrito Retiro")
    if superficie > 100:
        precio = 1000
    else:
        precio = 500
        
    print(f"El precio de la casa es: {precio}")
    
else:
    print("Otro distrito")
```

## Ejercicio 6
Dada la siguiente lista:

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que los recorra e imprima por pantalla todos los divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

for numero in list1:
    # Aqui si es mayor que 150 lo rompo
    if numero > 150:
        break
    
    # aqui dividimos por cero, asi comprobamos 
    if numero % 5 == 0:
        print(numero)
```

## Ejercicio 9
En este ejercicio vamos a crear un pequeño juego. Se trata de intentar adivinar un numero del 1 al 5. Tenemos dos intentos para acertar. Pasos a seguir:

1. Ya viene implementado con la librería `random` cómo obtener un número aleatorio del 1 al 5
2. Tendrás que declarar en una variable el numero de vidas, y mediante un bucle while, comprobar que todavia quedan vidas.
3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas.
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose".

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas.

```python
import random

# 1. numero se secreo
numero_secreto = random.randint(1, 5)

# 2. las vidas que tiene
vidas = 2

print("¡Bienvenido al juego! Adivina un número del 1 al 5.")

# 3. 
while vidas > 0:
   
    intento = int(input(f"Introduce tu número (Vidas restantes: {vidas}): "))
    
   
    if intento == numero_secreto:
        print("You win")
        break  
    else:
     
        vidas -= 1

# 4. 
else:
    print("You lose")
    print(f"El número secreto era: {numero_secreto}")
```
