### 2. Filtrar números pares de una lista:

Emplea filter para obtener sólo los números pares de la lista [1, 2, 3, 4, 5, 6].

Solución esperada: [2, 4, 6]

```python
lista = [1, 2, 3, 4, 5, 6]
numeros_pares = list(filter(lambda x:x % 2 ==0, lista ))
print ("Los numeros pares que arroja la funcion son", numeros_pares)
```

### 3. Convertir grados Celsius a Fahrenheit:

Usa map y una función lambda para convertir una lista de temperaturas en grados Celsius [0, 10, 20, 30] a Fahrenheit.

Solución esperada: [32.0, 50.0, 68.0, 86.0]

```python
celsius = [0,10,20,30,40]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print ("La temperatura en fahrenheit es:", fahrenheit)
```

### 4. Sumar los cuadrados de los números pares de una lista:

Combina filter, map y reduce (importa reduce desde functools) para sumar los cuadrados de los números pares en [1, 2, 3, 4, 5, 6].

Solución esperada: 56

```python
from functools import reduce

numeros = [1, 2, 3, 4, 5, 6]

# 1. Filtramos los números pares 
pares = filter(lambda x: x % 2 == 0, numeros)

# 2. Elevamos al cuadrado cada par 
cuadrados_pares = map(lambda x: x**2, pares)

# 3. Sumamos todos los cuadrados con reduce 
suma_cuadrados = reduce(lambda acumulador, valor: acumulador + valor, cuadrados_pares)

print("Suma de los cuadrados de los números pares:", suma_cuadrados)
```

### 5. Encadenar map, filter, y reduce para resolver un problema complejo:

Dada una lista de strings que representan números enteros y "NA" (por ejemplo, ["2", "5", "NA", "9", "10"]), convierte los números a enteros, filtra los "NA", suma el cuadrado de los números que son mayores que 5 y menores que 10.

Solución esperada: Un único valor que es la suma de los cuadrados de los números válidos.

```python
from functools import reduce

datos = ["2", "5", "NA", "9", "10"]

# 1. Filtramos los valores NA
sin_na = filter(lambda x: x != "NA", datos)

# 2. Convertimos los strings a enteros, el map espera la función y el iterable
enteros = map(int, sin_na)

# 3. Filtramos los números mayores que 5 y menores que 10 (5 < x < 10)
filtrados = filter(lambda x: 5 < x < 10, enteros)

# 4. Elevamos al cuadrado los números que quedan
cuadrados = map(lambda x: x**2, filtrados)

# 5. Sumamos todos los cuadrados con reduce
resultado = reduce(lambda acumulador, valor: acumulador + valor, cuadrados, 0)

print("Resultado final:", resultado)
```
