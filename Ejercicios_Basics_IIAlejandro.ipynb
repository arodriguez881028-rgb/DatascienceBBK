## Ejercicio 9
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965
4. Valor absoluto de: -74

````python

lista = [4, 6, 8, -1]

# 1. 
maximo = max(lista)
print(f"1. El máximo de la lista es: {maximo}")  # Salida: 8

# 2. sumar todo con sum
suma_total = sum(lista)
print(f"2. La suma de los elementos es: {suma_total}")  # Salida: 17

# 3. Redondea un float a 3 roud(3)
numero_decimal = 63.451256965
redondeado = round(numero_decimal, 3)
print(f"3. Número redondeado a 3 decimales: {redondeado}")  # Salida: 63.451

# 4. Calcula el valor absoluto utilizando abs()
valor_absoluto = abs(-74)
print(f"4. El valor absoluto de -74 es: {valor_absoluto}")  # Salida: 74
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python


texto = "Lo primero de todo, cómo están los máquinas"

# 1. Pasar a mayusculas
texto_mayus = texto.upper()
print(f"1. Mayúsculas: {texto_mayus}")

# 2. Pasarlo todo a minúsculas con lower()
texto_minus = texto.lower()
print(f"2. Minúsculas: {texto_minus}")

# 3. Pasar a mayúsculas solo las iniciales de las palabras con title()
texto_title = texto.title()
print(f"3. Iniciales en mayúscula: {texto_title}")

# 4. Crear una lista dividiéndolo por sus espacios con split()
lista_palabras = texto.split()
print(f"4. Lista por espacios: {lista_palabras}")

# 5. Sustituir las comas ',' por puntos y comas ';' con replace()
texto_reemplazado = texto.replace(",", ";")
print(f"5. Comas por puntos y comas: {texto_reemplazado}")

# 6. Eliminar las 'a' minúsculas reemplazándolas por un string vacío con replace()
texto_sin_a = texto.replace("a", "")
print(f"6. Sin 'a' minúsculas: {texto_sin_a}")
````
