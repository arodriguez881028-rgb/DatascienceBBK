9. ¿Cómo se llamaba la persona de mayor edad que murió en el Titanic? ¿Viajaba sola o tenía familiares a bordo?

Para responder a la segunda pregunta, averigua qué columnas contienen esa información. Como analista de datos, parte del trabajo consiste en conocer el conjunto de datos. En este caso, las columnas relevantes son:

- `sibsp`: número de hermanos o cónyuges a bordo.
- `parch`: número de padres o hijos a bordo.

```python
#Filtramos las filas de los pasajeros que no sobrevivieron
muertos = titanic[titanic['survived'] == 0]

# 2. Buscamos el índice de la persona de mayor edad entre los fallecidos
idx_mayor = muertos['age'].idxmax()

# 3. Extraemos toda la información de esa persona
persona_mayor = muertos.loc[idx_mayor]

print(persona_mayor[['name', 'age', 'sibsp', 'parch']])
```
