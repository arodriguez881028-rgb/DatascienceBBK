## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def dia_semana(numero):
    
    dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
    
    if 1 <= numero <= 7:
        return dias[numero - 1]
    else:
        return "Error: El número debe estar entre 1 y 7."

# Pruebas de la función
print("Ejercicio 1 (Día 3):", dia_semana(3))
print("Ejercicio 1 (Día 7):", dia_semana(7))
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python

def contar_letra(texto, letra):
    """
    Cuenta cuántas veces aparece una letra en un texto, 
    sin distinguir entre mayúsculas y minúsculas.
    """
    # pasamos todo a minuculas
    return texto.lower().count(letra.lower())


texto_ejemplo = "Programar en Python es bastante aburrido"
letra_a_buscar = "p"

resultado = contar_letra(texto_ejemplo, letra_a_buscar)

print(f"Texto: '{texto_ejemplo}'")
print(f"La letra '{letra_a_buscar}' aparece {resultado} veces.")
print("_" * 50)
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
def contar_letras_texto(texto):
    """
    Cuenta la frecuencia de cada letra en un string,
    unificando mayúsculas/minúsculas e ignorando espacios o puntuación.
    """
    conteo = {}
    
    texto_limpio = texto.lower()
    
    for caracter in texto_limpio:
        
        if caracter.isalpha():
            if caracter in conteo:
                conteo[caracter] += 1
            else:
                conteo[caracter] = 1
                
    return conteo


frase_ejemplo = "Programar en Python mola mucho"
resultado_conteo = contar_letras_texto(frase_ejemplo)

print("Ejercicio 5 - Frecuencia de letras:")
print(resultado_conteo)
print("_" * 50)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python

def fibonacci(n):
  
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        
        return fibonacci(n - 1) + fibonacci(n - 2)


print("Ejercicio 8 - Pruebas de la serie de Fibonacci:")
print(f"fibonacci(2) = {fibonacci(2)}")
print(f"fibonacci(4) = {fibonacci(4)}")
print(f"fibonacci(5) = {fibonacci(5)}")
print(f"fibonacci(8) = {fibonacci(8)}")
print("_" * 50)
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el área de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python

import math

def area_cuadrado(lado):
    """Calcula y devuelve el área de un cuadrado (lado al cuadrado)."""
    return lado ** 2

def area_triangulo(base, altura):
    """Calcula y devuelve el área de un triángulo ((base * altura) / 2)."""
    return (base * altura) / 2

def area_circulo(radio):
    """Calcula y devuelve el área de un círculo (pi * radio al cuadrado)."""
    return math.pi * (radio ** 2)
````

````python

parte_1 = (2 * area_circulo(10)) + area_triangulo(3, 7)

# 2. Un cuadrado de lado = 10 + 3 círculos (uno de radio 4 y dos de radio 6) + 5 triángulos de base 2 y altura 4
parte_2 = (
    area_cuadrado(10) + 
    area_circulo(4) + (2 * area_circulo(6)) + 
    (5 * area_triangulo(2, 4))
)

print("Ejercicio 9 - Resultados:")
print(f"Resultado del apartado 1: {parte_1:.2f}")
print(f"Resultado del apartado 2: {parte_2:.2f}")
print("_" * 50)
````
